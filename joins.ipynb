```python
data1 = [(1, "Alice"), (2, "Bob"), (3, "Charlie")]
data2 = [(2, "Finance"), (3, "Marketing"), (4, "Sales")]

df1 = spark.createDataFrame(data1, ["id", "name"])
df2 = spark.createDataFrame(data2, ["id", "department"])

df1.show()
df2.show()
```

```python
df=df1.join(df2, "id", "inner")
df.show()
```
