```python
%sql
-- Q1. Assign a unique row number to every employee ordered by salary from highest to lowest.
SELECT
  emp_id,
  name,
  department,
  salary,
  hire_date,
  manager_id,
  ROW_NUMBER() OVER (ORDER BY salary DESC) AS row_num
FROM practice1.test.emp
ORDER BY salary DESC
```

```python
%sql
--Q2. Within each department, assign a row number to employees ordered by salary descending (1 = highest earner in dept)
SELECT
  emp_id,
  name,
  department,
  salary,
  hire_date,
  manager_id,
  ROW_NUMBER() OVER (PARTITION BY department ORDER BY salary DESC) AS dept_row_num
FROM practice1.test.emp
ORDER BY department, dept_row_num
```

```python
%sql
--Q3. Show each employee's salary and the total salary bill for the entire company alongside every row.
SELECT
  emp_id,
  name,
  department,
  salary,
  (SELECT SUM(salary) FROM practice1.test.emp) AS total_salary_bill
FROM practice1.test.emp
```

```python
%sql
--Q4. For each sale, compute a running total of sale amounts per employee ordered by sale_date
SELECT
  sale_id,
  emp_id,
  sale_date,
  amount,
  region,
  SUM(amount) OVER (PARTITION BY emp_id ORDER BY sale_date ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS running_total
FROM practice1.test.sales
ORDER BY emp_id, sale_date
```

```python
%sql
--Q5. Show each employee's salary as a percentage of their department's total salary, rounded to 2 decimal places.
SELECT
  emp_id,
  name,
  department,
  salary,
  ROUND(
    salary * 100.0 / SUM(salary) OVER (PARTITION BY department),
    2
  ) AS salary_pct_of_dept
FROM practice1.test.emp
ORDER BY department, salary_pct_of_dept DESC
```

```python
%sql
--Q6. Rank employees by salary within each department
SELECT
  emp_id,
  name,
  department,
  salary,
  RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS dept_salary_rank
FROM practice1.test.emp
ORDER BY department, dept_salary_rank
```

```python
%sql
--Q7. Find employees whose department salary rank is exactly 2 — second highest in their dept.
SELECT
  emp_id,
  name,
  department,
  salary,
  RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS dept_salary_rank
FROM practice1.test.emp
WHERE RANK() OVER (PARTITION BY department ORDER BY salary DESC) = 2
ORDER BY department, salary DESC
```
